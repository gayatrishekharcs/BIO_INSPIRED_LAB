```python

import numpy as np
import random
import math

# 1. DEFINE THE DATA CENTER PROBLEM

heat_loads = [4, 6, 8]

min_temp = 18
max_temp = 27

number_of_zones = len(heat_loads)

lower_bounds = np.array([0.3, 18] * number_of_zones)
upper_bounds = np.array([1.0, 24] * number_of_zones)


# 2. INITIALIZE PARAMETERS

n_nests = 25
max_iterations = 100

pa = 0.25
alpha = 0.05
lam = 1.5


# 3. FITNESS FUNCTION

def fitness(solution):

    total_energy = 0
    penalty = 0

    for i in range(number_of_zones):

        airflow = solution[2 * i]
        setpoint = solution[2 * i + 1]

        load = heat_loads[i]

        server_temp = (
            29 + 0.65 * load
            - 9 * airflow
            + 0.35 * (setpoint - 20)
        )

        energy = (
            1.5 + 7 * airflow ** 3
            + 0.25 * (24 - setpoint)
        )

        total_energy += energy

        if server_temp > max_temp:
            penalty += 1000 * (server_temp - max_temp) ** 2

        if server_temp < min_temp:
            penalty += 1000 * (min_temp - server_temp) ** 2

    return total_energy + penalty


# 4. INITIALIZE POPULATION

nests = np.random.uniform(
    lower_bounds,
    upper_bounds,
    (n_nests, len(lower_bounds))
)

best_nest = None
best_fitness = float("inf")


# 5. LEVY FLIGHT

def levy_flight():

    sigma = (
        math.gamma(1 + lam)
        * math.sin(math.pi * lam / 2)
        /
        (
            math.gamma((1 + lam) / 2)
            * lam
            * 2 ** ((lam - 1) / 2)
        )
    ) ** (1 / lam)

    u = np.random.normal(0, sigma, len(lower_bounds))
    v = np.random.normal(0, 1, len(lower_bounds))

    step = u / (np.abs(v) ** (1 / lam) + 1e-10)

    return step


# 6. CUCKOO SEARCH ALGORITHM

for iteration in range(max_iterations):

    for i in range(n_nests):

        # Generate new solution using Levy flight

        new_nest = nests[i] + alpha * levy_flight() * (
            upper_bounds - lower_bounds
        )

        new_nest = np.clip(
            new_nest,
            lower_bounds,
            upper_bounds
        )

        j = random.randint(0, n_nests - 1)

        if fitness(new_nest) < fitness(nests[j]):
            nests[j] = new_nest

    # Evaluate all nests

    fitness_values = np.array([fitness(n) for n in nests])

    # Abandon worst nests

    n_abandon = max(1, int(pa * n_nests))

    worst_indices = np.argsort(fitness_values)[-n_abandon:]

    for index in worst_indices:

        nests[index] = np.random.uniform(
            lower_bounds,
            upper_bounds
        )

    # Find best nest

    fitness_values = np.array([fitness(n) for n in nests])

    best_index = np.argmin(fitness_values)

    if fitness_values[best_index] < best_fitness:

        best_fitness = fitness_values[best_index]
        best_nest = nests[best_index].copy()

    print(
        "Iteration:", iteration + 1,
        "| Best Fitness:", round(best_fitness, 4)
    )


# 7. DISPLAY BEST SOLUTION

print("\n====================================")
print(" DATA CENTER COOLING OPTIMIZATION")
print("====================================")

total_energy = 0

for i in range(number_of_zones):

    airflow = best_nest[2 * i]
    setpoint = best_nest[2 * i + 1]

    load = heat_loads[i]

    server_temp = (
        29 + 0.65 * load
        - 9 * airflow
        + 0.35 * (setpoint - 20)
    )

    energy = (
        1.5 + 7 * airflow ** 3
        + 0.25 * (24 - setpoint)
    )

    total_energy += energy

    print("\nServer Zone:", i + 1)
    print("Heat Load:", load, "kW")
    print("Airflow Rate:", round(airflow, 3))
    print("Temperature Setting:", round(setpoint, 2), "C")
    print("Server Temperature:", round(server_temp, 2), "C")
    print("Cooling Energy:", round(energy, 3))

    if min_temp <= server_temp <= max_temp:
        print("Status: SAFE")
    else:
        print("Status: UNSAFE")

print("\n====================================")
print("Minimum Fitness:", round(best_fitness, 4))
print("Total Cooling Energy:", round(total_energy, 4))
print("Optimization Completed!")
print("====================================")
```
